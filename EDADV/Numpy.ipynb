## Comparision of performance with list

```python
%%time
## method 1 for squaring the element of list using loop
N=1000000
list_ = list(range(N)) ## range(start, end, step)
for i in range(N):
    list_[i]=list_[i]*list_[i]

```

```python
%%time
# method 2 for squaring the element of list using list comprenhsive
N=1000000
list_ = list(range(N))
list_sq = [x*x for x in list_]
```

```python
# inline fucntion
square = lambda x: x*x
print(square(12))
```

```python
def square(x):
    return x*x

print(square(12))
```

```python
%%time
# method 3 for squaring the element of list using list comprenhsive
N=1000000
list_ = list(range(N))
list_sq = map(lambda x: x*x, list_)
```
