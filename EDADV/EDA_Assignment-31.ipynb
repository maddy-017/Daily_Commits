**(ii) Why standardise `gender` before removing duplicates?** Pandas compares text exactly. `'MALE '` and `'male'` look the same to us, but they are different strings, so two copies of the same patient written differently would **not** be found as duplicates. Justification: why did you choose that statistic rather than the other one? Use evidence from (a). Also mention one weakness of giving every patient the *same* replacement value (think about a 5-year-old and a 70-year-old).

**Why this matters**
`bmi` is missing for about 14% of patients, the largest gap in the dataset. Dropping them all would waste a lot of data, so you need a defensible fill value.

**Expected output / answer**
Descriptive statistics, before/after missing counts, and a justified choice (4-5 sentences).

> **Hint:** Look at the maximum and minimum of `bmi`, and compare mean with median. Which statistic is pulled around by extreme values?

```python
# (a) Explore bmi
print(df_clean["bmi"].describe())
bmi_mean = df_clean["bmi"].mean()
bmi_median = df_clean["bmi"].median()
print("\nMean  :", round(bmi_mean, 2))
print("Median:", round(bmi_median, 2))

# Extra evidence: does bmi depend on age? (checked BEFORE filling)
age_band = pd.cut(df_clean["age"], bins=[0, 18, 40, 60, 100], right=False)
print("\nbmi by age band (before filling):")
print(df_clean.groupby(age_band, observed=True)["bmi"].agg(["count", "mean", "median"]).round(2))

# (b) Fill the missing values and show the count before and after
print("\nMissing bmi before:", df_clean["bmi"].isnull().sum())
df_clean["bmi"] = df_clean["bmi"].fillna(bmi_median)
print("Missing bmi after :", df_clean["bmi"].isnull().sum())
```

A few very high BMI values pull the mean up (the mean is a little above the median), while the median is not affected by extreme values. The median is also a real, typical value in the data, so it is the safer choice.

**Weakness of one value for everyone:** BMI depends on age. In the age-band table, children under 18 have a median BMI of about **19.4**, while patients aged 40-59 have a median of about **30.7**. Filling every missing value with 28.2 makes missing children look too heavy and missing middle-aged patients look too light. It also creates a spike at 28.2: 167 rows (13.9%) now have exactly the same value, which slightly reduces the spread (std). A better method would be to fill with the median of each age group.

**Report for the hospital manager**

We studied the records of 1,200 patients (after removing 60 repeated rows). Of these, 75 patients (6.25%) had a stroke.

**What we found**
1. **Age is the clearest pattern.** Patients aged 60 or above had a stroke rate of 15.8%, compared with 2.8% for younger patients. None of the 213 children had a stroke.
2. **Heart disease and high blood pressure go with more strokes.** The stroke rate was 20.3% for patients with heart disease (5.5% without) and 15.3% for patients with high blood pressure (5.1% without).
3. **Smoking looked risky at first, but this was mostly because of age.** Former smokers were older (average 54.9 years). Among seniors, smokers (14.3%) and non-smokers (15.9%) had similar rates.

**Limitations**
- The data is a one-time snapshot from one hospital, so it shows links, not causes, and the 6.25% cannot be used for the whole population.
- Many values were missing or unknown (BMI for 14% of patients), and some groups are small (only 64 patients with heart disease), so some rates may change a lot with a few patients.

**Recommended next step:** Record BMI, smoking and job details for every patient at registration, give each patient a unique ID, and repeat this analysis with more patients.
